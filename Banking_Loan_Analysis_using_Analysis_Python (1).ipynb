```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/banking_loan_analysis.csv")
```

```python
print("First 5 Rows of Dataset:")
print(df.head())
```

```python
print("\nDataset Information:")
print(df.info())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df.fillna(method='ffill', inplace=True)
```

```python
# TASK 1 : Loan Approval Analysis
sns.countplot(x='Loan_Status', data=df)
plt.title("Loan Approval Status")
plt.show()
```

```python
# TASK 2 : Income Analysis
sns.histplot(df['ApplicantIncome'], kde=True)
plt.title("Applicant Income Distribution")
plt.show()
```

```python
# TASK 3 : Risk Profiling
sns.countplot(
    x='Credit_History',
    hue='Loan_Status',
    data=df
)
plt.title("Credit History vs Loan Status")
plt.show()
```

```python
#TASK 4 : Approval Trends
sns.countplot(
    x='Education',
    hue='Loan_Status',
    data=df
)
plt.title("Education vs Loan Approval")
plt.show()
```

```python
df['Loan_Status'] = df['Loan_Status'].map({'Y':1, 'N':0})
numeric_df = df.select_dtypes(include=['int64', 'float64'])
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
print("\nProject Insights:")
print("1. Customers with good credit history get more loan approvals.")
print("2. Higher income applicants have better approval chances.")
print("3. Education affects loan approval trends.")
```
